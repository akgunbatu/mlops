```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

**a)**

```python
n = 1000000
sh = stats.expon.rvs(scale = 10, size = n)
extract = stats.norm.rvs(15,3,n)
devexp = stats.norm.rvs(3,0.5,n)
devmodel = stats.poisson.rvs(mu = 30, size = n)
process = stats.norm.rvs(50,20,n)
datapipe = stats.poisson.rvs(mu = 100, size = n)
model = stats.norm.rvs(30,5,n)
```

**b1)**

```python
days = sh + extract+ devexp + devmodel + process + datapipe + model
days
```

```python
print(np.mean(days), np.median(days), np.quantile(days,0.95))
```

**b2)**

```python
print("SH Analysis:", np.mean(sh))
print("Data Extract:", np.mean(extract))
print("Dev Exploration:", np.mean(devexp))
print("Dev Modeling:", np.mean(devmodel))
print("Process Int:", np.mean(process)) 
print("Data Pipeline:", np.mean(datapipe)) 
print("Model Productionization:", np.mean(model))
```

```python
print("SH Analysis:", np.std(sh))
print("Data Extract:", np.std(extract))
print("Dev Exploration:", np.std(devexp))
print("Dev Modeling:", np.std(devmodel))
print("Process Int:", np.std(process)) 
print("Data Pipeline:", np.std(datapipe)) 
print("Model Productionization:", np.std(model))
```

**The data pipeline has the highest mean, so it contributes the most to the overall time. The process integration has the highest standard deviation, so it contributes the most to the uncertainty to the overall time.**

```python
datapipe_new = stats.poisson.rvs(mu = 150, size = n)
days_new = sh + extract+ devexp + devmodel + process + datapipe_new + model
days_new
```

```python
delay = days_new - days
print(np.mean(delay), np.quantile(delay,0.95))
```
